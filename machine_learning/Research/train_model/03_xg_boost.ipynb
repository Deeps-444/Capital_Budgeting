# XG boost

```python
import pandas as pd
from pathlib import Path

data_path = Path("../data/raw_dataset.csv")

df = pd.read_csv(data_path)

print("Dataset shape:", df.shape)

df.head()
```

```python
target = "cash_flow"

features = [
    "project_type",
    "project_scale",
    "project_duration_years",
    "complexity_index",
    "initial_investment",
    "tax_rate",
    "risk_index",
    "year",
    "inflation_rate",
    "market_growth_rate"
]

X = df[features]
y = df[target]
```

```python
from sklearn.model_selection import train_test_split

project_ids = df["project_id"].unique().to_numpy()

train_projects, test_projects = train_test_split(
    project_ids,
    test_size=0.20,
    random_state=42
)

train_df = df[
    df["project_id"].isin(train_projects)
].copy()

test_df = df[
    df["project_id"].isin(test_projects)
].copy()

print("Training projects:", train_df["project_id"].nunique())
print("Testing projects:", test_df["project_id"].nunique())

print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))
```

```python
overlap = (
    set(train_df["project_id"])
    & set(test_df["project_id"])
)

print("Projects appearing in both sets:", len(overlap))
```

```python
X_train = train_df[features]
y_train = train_df[target]

X_test = test_df[features]
y_test = test_df[target]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
categorical_features = [
    "project_type",
    "project_scale"
]

numerical_features = [
    "project_duration_years",
    "complexity_index",
    "initial_investment",
    "tax_rate",
    "risk_index",
    "year",
    "inflation_rate",
    "market_growth_rate"
]
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)
```

```python
from sklearn.pipeline import Pipeline

xgb_pipeline = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        ("model", xgb_model)
    ]
)
```

```python
xgb_pipeline.fit(
    X_train,
    y_train
)
```

```python
y_pred = xgb_pipeline.predict(X_test)

print("Number of predictions:", len(y_pred))
```

```python
comparison = pd.DataFrame({
    "actual_cash_flow": y_test.values,
    "predicted_cash_flow": y_pred
})

comparison.head(10)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = mean_squared_error(
    y_test,
    y_pred
) ** 0.5

r2 = r2_score(
    y_test,
    y_pred
)

print(f"MAE:  ₹{mae:,.2f}")
print(f"RMSE: ₹{rmse:,.2f}")
print(f"R²:   {r2:.4f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 7))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.6
)

plt.xlabel("Actual Cash Flow")
plt.ylabel("Predicted Cash Flow")
plt.title("XG Boost: Actual vs Predicted Cash Flow")

plt.show()
```

feature importance

```python
model = xgb_pipeline.named_steps["model"]

preprocessor_fitted = (
    xgb_pipeline
    .named_steps["preprocessing"]
)

feature_names = (
    preprocessor_fitted
    .get_feature_names_out()
)
```

```python
feature_importance = pd.DataFrame({
    "feature": feature_names,
    "importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    "importance",
    ascending=False
)

feature_importance
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    feature_importance["feature"],
    feature_importance["importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("XG Bosst  Feature Importance")

plt.gca().invert_yaxis()

plt.show()
```
